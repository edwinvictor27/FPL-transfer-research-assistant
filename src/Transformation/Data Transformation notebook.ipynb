```python
cte_1 = (
    fixture_player_stats
    .groupBy("fixture_id", "stat_type", "player_id")
    .agg(
        first("value").alias("value"),
        count("*").alias("cnt")
    )
    .orderBy(
        col("fixture_id").asc(),
        col("player_id").asc(),
        col("stat_type").asc(),
        col("cnt").asc()
    )
)

# CTE 2
cte_2 = fpl_player_data.select("id", "web_name")

# Final query
result = (
    cte_1.alias("c1")
    .join(
        cte_2.alias("c2"),
        col("c1.player_id") == col("c2.id"),
        "left"
    )
    .filter(col("c1.fixture_id") == 1)
    .select("c2.*", "c1.*")
    .orderBy(
        col("c1.fixture_id").asc(),
        col("c1.player_id").asc(),
        col("c1.stat_type").asc(),
        col("c1.cnt").asc()
    )
)
```
